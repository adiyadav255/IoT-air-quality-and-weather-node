```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras

from google.colab import drive
drive.mount('/content/drive')

X_train = pd.read_csv('/content/drive/My Drive/ML_Data/X_train_scaled.csv')
X_test = pd.read_csv('/content/drive/My Drive/ML_Data/X_test_scaled.csv')
y_train = pd.read_csv('/content/drive/My Drive/ML_Data/y_train.csv')
y_test = pd.read_csv('/content/drive/My Drive/ML_Data/y_test.csv')

X_train = X_train.values
X_test  = X_test.values
y_train = y_train.values
y_test  = y_test.values

```

```python
from sklearn.preprocessing import StandardScaler

y_train = y_train.reshape(-1, 1)
y_test  = y_test.reshape(-1, 1)

y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train)
y_test_scaled  = y_scaler.transform(y_test)
```

Window Creation

```python
def create_windowed_data(X, y, window_size=8):
    Xw, yw = [], []
    for i in range(len(X) - window_size):
        Xw.append(X[i:i+window_size])
        yw.append(y[i+window_size])
    return np.array(Xw), np.array(yw)

WINDOW = 8

X_train_w, y_train_w = create_windowed_data(X_train, y_train_scaled, window_size=WINDOW)
X_test_w,  y_test_w  = create_windowed_data(X_test,  y_test_scaled,  window_size=WINDOW)

print("Windowed Shapes:")
print("X_train_w:", X_train_w.shape)
print("y_train_w:", y_train_w.shape)
print("X_test_w :", X_test_w.shape)
print("y_test_w :", y_test_w.shape)
```

Building a very small MLP for a TinyNN

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(WINDOW, 4)),
    keras.layers.Dense(16, activation='relu'),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

model.summary()
```

Model Training

```python
history = model.fit(
    X_train_w, y_train_w,
    validation_data=(X_test_w, y_test_w),
    epochs=40,
    batch_size=32,
    verbose=1
)
```

Make predictions on the test set and then visualize them against the actual values.

```python
import matplotlib.pyplot as plt

# Make predictions on the test data
y_pred_w = model.predict(X_test_w)
y_test_real = y_scaler.inverse_transform(y_test_w).reshape(-1)
y_pred_real = y_scaler.inverse_transform(y_pred_w).reshape(-1)
# Plot actual vs. predicted values
plt.figure(figsize=(12, 6))
plt.plot(y_test_real, label='Actual Values')
plt.plot(y_pred_real, label='Predicted Values')
plt.title('Model Predictions vs. Actual Values')
plt.xlabel('Time Step / Sample Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
y_test_real = y_scaler.inverse_transform(y_test_w)
y_pred_real = y_scaler.inverse_transform(y_pred_w)

plt.figure(figsize=(8, 8))
plt.scatter(y_test_real, y_pred_real, alpha=0.3)
min_val = min(y_test_real.min(), y_pred_real.min())
max_val = max(y_test_real.max(), y_pred_real.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', label="Ideal") # Diagonal line
plt.title('Actual vs. Predicted Values (Scatter Plot)')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.grid(True)
plt.show()
```

Final loss and MAE.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

y_pred_scaled = model.predict(X_test_w)
y_pred_real = y_scaler.inverse_transform(y_pred_scaled).reshape(-1)
y_test_real = y_scaler.inverse_transform(y_test_w).reshape(-1)

mae_real = mean_absolute_error(y_test_real, y_pred_real)
rmse_real = np.sqrt(mean_squared_error(y_test_real, y_pred_real))

print(f"Real AQI MAE : {mae_real:.3f}")
print(f"Real AQI RMSE: {rmse_real:.3f}")
```

Visualize the training and validation loss and MAE over epochs

```python
import matplotlib.pyplot as plt

# Plot training & validation loss values
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(loc='upper right')

# Plot training & validation MAE values
plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Train MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.title('Model Mean Absolute Error')
plt.ylabel('MAE')
plt.xlabel('Epoch')
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

Export Model

```python
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

with open("tinyNN_float32.tflite", "wb") as f:
    f.write(tflite_model)

print("Saved: tinyNN_float32.tflite")
```

```python
def representative_data_gen():
    for i in range(200):
        idx = np.random.randint(0, len(X_train_w))
        yield [X_train_w[idx:idx+1].astype(np.float32)]

converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = representative_data_gen
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type = tf.int8
converter.inference_output_type = tf.int8

tflite_quant = converter.convert()

with open("tinyNN_int8.tflite", "wb") as f:
    f.write(tflite_quant)

print("Saved: tinyNN_int8.tflite")
from google.colab import files
files.download("tinyNN_int8.tflite")
```
